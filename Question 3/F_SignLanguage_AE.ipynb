```python
import pandas as pd

train_df = pd.read_csv(f"{BASE}/AutoEncoder/Sign Language MNIST/sign_mnist_train.csv")
test_df = pd.read_csv(f"{BASE}/AutoEncoder/Sign Language MNIST/sign_mnist_test.csv")

print("Train shape : ", train_df.shape)
print("Test shape : ", test_df.shape)
print(sorted(train_df["label"].unique()))
print("Y rows in train : ", (train_df["label"] == 24).sum())
print("Y rows in test : ", (test_df["label"] == 24).sum())
print("Missing values in Train & Test set : ", train_df.isnull().sum().sum() + test_df.isnull().sum().sum())
```

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.metrics import confusion_matrix, classification_report
```

```python
X_train = train_df.drop("label", axis=1).values / 255.0
y_train = train_df["label"].values

X_test = test_df.drop("label", axis=1).values / 255.0
y_test = test_df["label"].values

X_train_normal = X_train[y_train != 24]
y_test_binary = (y_test == 24).astype(int)

print(X_train_normal.shape)
print(X_test.shape)
print(y_test_binary.sum())
```

```python
model = keras.Sequential([
    keras.Input(shape=(784,)),
    layers.Dense(128, activation="relu"),
    layers.Dense(32, activation="relu"),
    layers.Dense(128, activation="relu"),
    layers.Dense(784, activation="sigmoid"),
])

model.summary()
```

```python
model.compile(
    optimizer = keras.optimizers.Adam(learning_rate = 0.001),
    loss = "mse",
    metrics = ["mae"]
)
```

```python
history = model.fit(
    X_train_normal, X_train_normal,
    epochs = 20,
    batch_size = 32,
    validation_split = 0.1
)
```

```python
train_pred= model.predict(X_train_normal)
train_errors = np.mean((train_pred - X_train_normal) ** 2, axis = 1)

threshold = np.percentile(train_errors, 95)
print("Threshold : ", threshold)
```

```python
test_pred = model.predict(X_test)
test_errors = np.mean((test_pred - X_test) ** 2, axis = 1)

y_pred = (test_errors > threshold).astype(int)

print("Confusion Matrix : ", confusion_matrix(y_test_binary, y_pred))
print("Classification Report : ", classification_report(y_test_binary, y_pred, target_names=["Other letters", "Y"]))
```
